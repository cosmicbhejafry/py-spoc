````python

data_gen_name = 'GMM_isotropic_heteroskedastic_balanced'
category_lst = ['blobbiness']

random_seeds_lst = [4948,230]

n_samples_lst = [1000,10000]
n_features_lst = [10,100,1000,10000]
# define some number of clusters according to the number of features
# {num_feature : [list of num_cluster params to sweep over]}
num_clusters_dict = {10 : [3,5,7], 100 : [3,10], 1000 : [3,10,40], 10000 : [3,10,40]}

cluster_stds_log_mean = 0.0
# dict { num_cluster : [list of diff sigma vals to try for the lognormal]
cluster_sig_dict = {
    3 : [0.3,0.5],
    5 : [0.3,0.5],
    7 : [0.3,0.5],
    10 : [0.3,0.5],
    40 : [0.3,0.5]
}

for n_samples, n_features in product(n_samples_lst, n_features_lst):
    
    print(n_samples, n_features, X.shape)

    for n_centers in num_clusters_dict[n_features]:

        for cluster_sig_val in cluster_sig_dict[n_centers]:

            for sample_num,seed in enumerate(random_seeds_lst):
                
                cluster_std_seed = seed+4
                rng = np.random.default_rng(cluster_std_seed)
                cluster_std_lst = rng.lognormal(mean=0.0, sigma=cluster_sig_val, size=n_centers).tolist()
                
                X, y = datasets.make_blobs(n_samples=n_samples,n_features=n_features, random_state=seed, centers=n_centers,cluster_std=cluster_std_lst)
                
                params_dict = {
                    'num_clusters': n_centers,
                    'cluster_std_sig' : cluster_sig_val,
                    'cluster_std_seed' : cluster_std_seed,
                    'cluster_std_min_max' : (min(cluster_std_lst),max(cluster_std_lst)),
                    'cluster_std_method' : 'seeded, generated using a lognormal distribution , mu = 0, sig = cluster_std_sig'
                }
                
                extra_mtx_params = {
                    'cluster_labels' : y,
                    'cluster_std_lst' : cluster_std_lst
                }
                
                try:
                    save_dataset(X, 
                                generator=data_gen_name,
                                params=params_dict,
                                seed=seed,
                                category=category_lst,
                                extra_arrays=extra_mtx_params, 
                                root=DATA_ROOT)

                except:
                    print('error in saving')
````

## heteroskedastic, imbalanced clusters

### below code uses lognormal dist to get a list of cluster standard deviations, for heteroskedasticity

going to be noisy / lead to overlapping clusters

vary sigma

```
rng = np.random.default_rng(seed)
cluster_std = rng.lognormal(mean=0.0, sigma=0.5, size=n_centers).tolist()
```

### use power law relationshp to find cluster size proportions

$$w_k = \dfrac{1}{k^s}, k = 1,2,3,..K$$

w_k = cluster size proportion for cluster k, K = total number of clusters, s = hyperparameter to vary

s = 0 is when w_k = 1 (balanced case)

normalize w_k's to get proportions

as s increases, w_0 >> w_1 >> w_2 >> ..

a power law reln. (ref https://en.wikipedia.org/wiki/Power_law)

````python
s = 0
n_centers = 3

def get_cluster_sizes_power_law(decay_exp, n_clusters, n_samples, min_count=20):
    """
    get decaying cluster imbalance
    currently relies on manual tuning of the decay parameter
    default: apply a min count per cluster of 20
    """
    if n_samples < min_count * n_clusters:
        raise ValueError(f"n_samples={n_samples} cannot give {n_clusters} clusters "
                         f"at least {min_count} points each")
        
    w = 1 / np.arange(1, n_clusters + 1) ** decay_exp
    w /= w.sum()
    spare = n_samples - min_count * n_clusters
    counts = min_count + np.floor(w * spare).astype(int)
    counts[0] += n_samples - counts.sum()
    return counts.tolist()
````

````python

# {num_cluster : [list of decay parameters]}
size_decay_dict = {
    3 : [0.5,1,2],
    10 : [0.5,1],
    40 : [0.5]
}

for nc in [3,10,40]:
    
    s_lst = size_decay_dict[nc]
            
    for s in s_lst:
        counts = get_cluster_sizes_power_law(s,nc,10000)
        print(max(counts)/min(counts))
        print(counts)
        print('---')
````

````python
data_gen_name = 'GMM_isotropic_heteroskedastic_imbalanced'
category_lst = ['blobbiness']

random_seeds_lst = [342,80]

n_samples_lst = [1000,10000]
n_features_lst = [10,100,1000,10000]
# define some number of clusters according to the number of features
# {num_feature : [list of num_cluster params to sweep over]}
num_clusters_dict = {
    10 : [3,5,7], 100 : [3,10], 
    1000 : [3,10,40], 10000 : [3,10,40]
}

# {num_cluster : [list of decay parameters]}
size_decay_dict = {
    3 : [0.5,2], 5 : [0.5,2], 7 : [0.5,2],
    10 : [0.5], 40 : [0.5]
}

# MORE EXTENDED, IF REQUIRED:
# size_decay_dict = {
#     3 : [0.5,1,2], 5 : [0.5,1,2], 7 : [0.5,1,2],
#     10 : [0.5,1], 40 : [0.5]
# }

cluster_stds_log_mean = 0.0
# dict { num_cluster : [list of diff sigma vals to try for the lognormal]
cluster_sig_dict = {
    3 : [0.3,0.5],
    5 : [0.3,0.5],
    7 : [0.3,0.5],
    10 : [0.3,0.5],
    40 : [0.3,0.5]
}


for n_samples, n_features in product(n_samples_lst, n_features_lst):
    
    print(n_samples, n_features, X.shape)

    for n_centers in num_clusters_dict[n_features]:
        
        for decay_exp in size_decay_dict[n_centers]:
            
            cluster_sizes = get_cluster_sizes_power_law(decay_exp,n_centers,n_samples)

            for cluster_sig_val in cluster_sig_dict[n_centers]:

                for sample_num,seed in enumerate(random_seeds_lst):
                    
                    cluster_std_seed = seed+4
                    rng = np.random.default_rng(cluster_std_seed)
                    cluster_std_lst = rng.lognormal(mean=cluster_stds_log_mean, sigma=cluster_sig_val, size=n_centers).tolist()
                    
                    X, y = datasets.make_blobs(n_samples=cluster_sizes,n_features=n_features, random_state=seed, centers=None,cluster_std=cluster_std_lst)
                    
                    params_dict = {
                        'num_clusters': n_centers,
                        'cluster_size_decay_param' : decay_exp,
                        'n_samples' : n_samples,
                        'cluster_size_method' : 'use the get_cluster_sizes_power_law method described in the data gen notebook',
                        'cluster_std_sig' : cluster_sig_val,
                        'cluster_std_logmean' : cluster_stds_log_mean,
                        'cluster_std_seed' : cluster_std_seed,
                        'cluster_std_min_max' : (min(cluster_std_lst),max(cluster_std_lst)),
                        'cluster_std_method' : 'seeded, generated using a lognormal distribution , mu = 0, sig = cluster_std_sig'                        
                    }
                    
                    extra_mtx_params = {
                        'cluster_labels' : y,
                        'cluster_sizes' : cluster_sizes,
                        'cluster_std_lst' : cluster_std_lst                        
                    }
                    
                    try:
                        save_dataset(X, 
                                    generator=data_gen_name,
                                    params=params_dict,
                                    seed=seed,
                                    category=category_lst,
                                    extra_arrays=extra_mtx_params, 
                                    root=DATA_ROOT)

                    except:
                        print('error in saving')
````
