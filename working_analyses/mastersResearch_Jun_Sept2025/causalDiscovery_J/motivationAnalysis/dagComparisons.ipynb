```python
from pyspoc import Calculator, Config
```

# Running reduced statistics
## 1) Distance Distribution Stats
### a) Euclidean distances; mean centroid; estimating number of peaks with KDE

```python
distance_distrib_stats_path = os.path.join(project_root,'hjResearch_2025','summary_stats','distance_distrib_stats.py')
print(distance_distrib_stats_path)
```

```python
kde_config_yaml = f"""
ReducedStatistics:
  {distance_distrib_stats_path}:
    DistanceDistributionModesSummarize:
      schemes:
        scheme_1:
          metric: euclidean
          point: global_mean_centroid
          type: kde
"""
kde_config_name = "kde_euclidean_globalmeancentroid"
cfg = Config.from_yaml(kde_config_name, kde_config_yaml)
```

```python
def process_motivation_dag(dag_number, project_root):

    base_dir = os.path.join(project_root, 'mastersResearch_Jun_Sept2025', 'causalDiscovery_J', 'motivationAnalysis')
    meta_file = os.path.join(base_dir, f'motivation_dag_{dag_number}.csv')
    
    print(f"--- Processing motivation_dag_{dag_number}.csv ---")
    meta_df = pd.read_csv(meta_file)
    display(meta_df)

    calc = Calculator(meta_df)
    calc.compute(cfg)
    results = calc.results
    print(results)

process_motivation_dag(1, project_root)
process_motivation_dag(2, project_root)
```

### b) Euclidean distances; mean centroid; mean

```python
euclideanMetric_meanCentroid_config_yaml = f"""
ReducedStatistics:
  {distance_distrib_stats_path}:
    DistanceDistributionBasicSummarize:
      schemes:
        scheme_1:
          metric: euclidean
          stats: 
            - mean
            - median
            - iqr
            - std
            - min
            - max
          point: global_mean_centroid
"""
euclideanMetric_meanCentroid_config_name = "kde_euclidean_globalmeancentroid"
cfg = Config.from_yaml(euclideanMetric_meanCentroid_config_name, euclideanMetric_meanCentroid_config_yaml )
```

```python
process_motivation_dag(1, project_root)
process_motivation_dag(2, project_root)
```
