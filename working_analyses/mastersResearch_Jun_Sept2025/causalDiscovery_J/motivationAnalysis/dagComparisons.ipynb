```python
def process_motivation_dag(dag_number, project_root):

    base_dir = os.path.join(project_root, 'mastersResearch_Jun_Sept2025', 'causalDiscovery_J', 'motivationAnalysis')
    meta_file = os.path.join(base_dir, f'motivation_dag_{dag_number}.csv')
    
    print(f"--- Processing motivation_dag_{dag_number}.csv ---")
    meta_df = pd.read_csv(meta_file)
    display(meta_df)

    calc = Calculator(meta_df)
    calc.compute(cfg)
    results = calc.results
    print(results)

process_motivation_dag(1, project_root)
process_motivation_dag(2, project_root)
```
