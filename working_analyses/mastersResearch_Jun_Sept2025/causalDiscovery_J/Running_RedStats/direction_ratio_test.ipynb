```python
from pyspoc import Calculator, Config
from pathlib import Path
import os
import numpy as np
import pandas as pd
```

```python
project_root = Path('./').resolve().parents[2]
print(project_root)
direction_ratio_path = os.path.join(project_root,'mastersResearch_Jun_Sept2025','causalDiscovery_J','ReducedStatistics','direction_ratio.py')
print(direction_ratio_path)
```

```python
direction_ratio_config_yaml = f"""
ReducedStatistics:
  {direction_ratio_path}:
    DirectedRatio:
        schemes:
            scheme_1:
                alpha: 0.05
"""
direction_ratio_config_name = "direction_ratio"
cfg = Config.from_yaml(direction_ratio_config_name, direction_ratio_config_yaml)
```

```python
base_dir = os.path.join(project_root, 'mastersResearch_Jun_Sept2025', 'causalDiscovery_J', 'motivationAnalysis')
meta_file1 = os.path.join(base_dir, f'motivation_dag_1.csv')

base_dir2 = os.path.join(project_root, 'mastersResearch_Jun_Sept2025', 'causalDiscovery_J', 'dataGen', 'loop100data')
meta_file2 = os.path.join(base_dir2, f'dag2dataset1.csv')

meta_df = pd.read_csv(meta_file2)
calc = Calculator(meta_df)
calc.compute(cfg)
results = calc.results
print(results)
```
