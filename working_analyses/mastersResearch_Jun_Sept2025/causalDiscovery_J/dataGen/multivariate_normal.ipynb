```python
import numpy as np
import pandas as pd

N = 10000
p = 10

mu = np.random.rand(p)
A = np.random.rand(p, p)
sigma = A @ A.T
scale_factor = 10000 # To make variances large
sigma_scaled = sigma * scale_factor

data = np.random.multivariate_normal(mu, sigma_scaled, N)
df = pd.DataFrame(data)

df.to_csv("multivariate_normal_data.csv", index=False)
print("Done")
```
